```python
import sys
import os

sys.path.append(os.path.abspath("../.."))

from config.spark import spark
```

```python
product_category_name_translation_df = (
    spark.read.format('csv')
    .option('header','True')
    .option('inferschema','True')
    .load('../../landing/files/product_category_name_translation.csv')
    
)
```

```python
from pyspark.sql.functions import current_timestamp

product_category_name_translation_df = product_category_name_translation_df.withColumn("ingestion_time",current_timestamp())
```

```python
product_category_name_translation_df.printSchema()
```

```python
product_category_name_translation_df.write.mode("overwrite").saveAsTable("bronze.product_category_name_translation")
```
